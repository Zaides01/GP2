```python
import pandas as pd
```

### Соединяем паралленьные таблицы

```python
wallget_0 = pd.read_csv('csv_files/posts.csv')
wallget_1 = pd.read_csv('csv_files/posts_1.csv')

df_wall = pd.concat([wallget_0, wallget_1], ignore_index=True)
df_wall = df_wall.drop(columns=['geo', 'zoom_text', 'author_ad', 'signer_id'], errors='ignore')
df_wall.shape
```

```python
videoget_0 = pd.read_csv('csv_files/videos.csv')
videoget_1 = pd.read_csv('csv_files/videos_1.csv')

df_video = pd.concat([videoget_0, videoget_1], ignore_index=True)
df_video.shape
```

```python
groupsget_0 = pd.read_csv('csv_files/groups.csv')
groupsget_1 = pd.read_csv('csv_files/groups_1.csv')

df_groups = pd.concat([groupsget_0, groupsget_1], ignore_index=True)
df_groups.shape
```

```python
df_groups.columns
```

```python
df_groups = df_groups.drop(columns = ['age_limits', 'is_closed', 'type'    , 'is_admin' , 'is_member', 'is_advertiser', 
                                      'photo_50'  , 'photo_100', 'contacts', 'photo_200', 'city'     ,], errors = 'ignore')
```

```python
df_groups.head()
```

```python
getcomments_0 = pd.read_csv('csv_files/comments.csv')
getcomments_1 = pd.read_csv('csv_files/comments_1.csv')

df_comments = pd.concat([getcomments_0, getcomments_1], ignore_index=True)
df_comments.shape
```

### Делаем два итоговых датасета

#### Видео

```python
df_video.columns
```

```python
df_video = df_video.drop(columns = ['player'], errors = 'ignore')
```

```python
df_video["owner_id"] = df_video["owner_id"].abs()
df_comments["owner_id"] = df_comments["owner_id"].abs()

df_video_comments = df_video.merge(df_comments, on=["owner_id", "id"], how="left")
```

```python
df_video_comments.head()
```

```python
df_full_video = df_video_comments.merge(df_groups, left_on="owner_id", right_on="id", how="left")
```

```python
df_full_video.to_csv("csv_files/df_video.csv", index=False, encoding="utf-8")
```

```python
df_full_video.shape
```

#### Стена

```python
df_wall.columns
```

```python
df_wall = df_wall.drop(columns = ['inner_type', 'push_subscription', 'hash'        , 'reaction_set_id', 'post_type',
                                  'edited'    , 'carousel_offset'  , 'copy_history', 'is_favorite'    , 'type'     ,
                                  'donut'     ,], errors = 'ignore')
```

```python
df_wall["owner_id"] = df_wall["owner_id"].abs()
df_wall_groups = df_wall.merge(df_groups, left_on="owner_id", right_on="id", how="left")
```

```python
df_wall_groups.shape
```

```python
df_wall_groups.to_csv("csv_files/df_wall.csv", index=False, encoding="utf-8")
```
