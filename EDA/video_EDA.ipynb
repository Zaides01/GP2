# Анализ и визуализация на таргет

```python
#наш таргет - количество просмотров
print(df['views'].isnull().sum())
print(df['views'].min())
print(df['views'].max())
```

Нет пропусков в кол-вах просмотров, мин 1, макс 190.031.826

```python
plt.figure(figsize=(10, 5))
plt.hist(df['views'],bins=75, color='skyblue')
plt.xlabel('Просмотры')
plt.ylabel('Frequency')
plt.title('Распределение просмотров видео')
plt.show()
```

Видим, что есть выбросы

```python
#до удаления выбросов
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
plt.boxplot(df['views'])
plt.ylabel('Просмотры')

plt.subplot(1, 2, 2)
plt.hist(df['views'], bins=75, color='skyblue')
plt.xlabel('Просмотры')
plt.ylabel('Frequency')
plt.title('Распределение просмотров видео')
plt.show()

Q1 = df['views'].quantile(0.25)
Q3 = df['views'].quantile(0.75)
IQR = Q3 - Q1

# Границы дл экстремальных выбросов
lower_line = Q1 - 3 * IQR
upper_line = Q3 + 3 * IQR

df_filtered = df[(df['views'] >= lower_line) & (df['views'] <= upper_line)]

# после удаления выбросов
plt.figure(figsize=(15, 5))

plt.subplot(1, 2, 1)
plt.boxplot(df_filtered['views'])
plt.ylabel('Просмотры')

plt.subplot(1, 2, 2)
plt.hist(df_filtered['views'], bins=75, color='skyblue')
plt.xlabel('Просмотры')
plt.ylabel('Frequency')
plt.title('Распределение просмотров видео')
plt.show()
```

```python
print(f"Изначальная статистика:\n{df['views'].describe()}")
print(f"\nСтатистика после удаления выбросов:\n{df_filtered['views'].describe()}")
```

```python
#Строим карту корреляции от количественных признаков
plt.figure(figsize=(10, 8))
correlation_matrix = df_filtered[['views',  'comments', 'duration', 'members_count', 'album_video_percent']].corr()
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Карта корреляции')
plt.show()
```

```python
#Строим графики рассеяния для признаков, влияющих на просмотры
plt.figure(figsize=(5, 5))
plt.scatter(df_filtered['duration'], df_filtered['views'],color='green', alpha=0.3)
plt.xlabel('Длительность')
plt.ylabel('Просмотры')
plt.title('Зависимость просмотров от длительности видео')
plt.show()

plt.figure(figsize=(5, 5))
plt.scatter(df_filtered['comments'], df_filtered['views'], alpha=0.3)
plt.xlabel('Комментарии')
plt.ylabel('Просмотры')
plt.title('Зависимость просмотров от кол-ва комментариев')
plt.show()

plt.figure(figsize=(5, 5))
plt.scatter(df_filtered['members_count'], df_filtered['views'],color='orange', alpha=0.3)
plt.xlabel('Подписчики')
plt.ylabel('Просмотры')
plt.title('Зависимость просмотров от кол-ва подписчиков')
plt.show()
```

**Графики получились скученные, нужно посмотреть на выбросы**

```python
plt.figure(figsize=(15, 5))
plt.hist(df['duration'], bins=75, color='green')
plt.xlabel('Длительность')
plt.ylabel('Frequency')
plt.title('Распределение длительности видео')
plt.show()

plt.figure(figsize=(15, 5))
plt.hist(df['comments'], bins=75, color='blue')
plt.xlabel('Кол-во комментариев')
plt.ylabel('Frequency')
plt.title('Распределение кол-ва комментариев')
plt.show()

plt.figure(figsize=(15, 5))
plt.hist(df['members_count'], bins=200, color='orange')
plt.xlabel('Подписчики')
plt.ylabel('Frequency')
plt.title('Распределение кол-ва подписчиков')
plt.show()
```

```python
#убираем выбросы в длительности в количестве комментариев
Q1_1 = df_filtered[['duration','comments']].quantile(0.25)
Q3_1 = df_filtered[['duration','comments']].quantile(0.75)
IQR_1 = Q3_1 - Q1_1

lower_line_1 = Q1_1 - 3 * IQR_1
upper_line_1 = Q3_1 + 3 * IQR_1

df_filter = df_filtered[
    (df_filtered['duration'] >= lower_line_1['duration']) &
    (df_filtered['duration'] <= upper_line_1['duration']) &
    (df_filtered['comments'] >= lower_line_1['comments']) &
    (df_filtered['comments'] <= upper_line_1['comments'])
]
```

```python
#еще раз строим графики рассеяния
plt.figure(figsize=(5, 5))
plt.scatter(df_filter['duration'], df_filter['views'],color='green', alpha=0.3)
plt.xlabel('Длительность')
plt.ylabel('Просмотры')
plt.title('Зависимость просмотров от длительности видео')
plt.show()

plt.figure(figsize=(5, 5))
plt.scatter(df_filter['comments'], df_filter['views'], alpha=0.3)
plt.xlabel('Комментарии')
plt.ylabel('Просмотры')
plt.title('Зависимость просмотров от кол-ва комментариев')
plt.show()

plt.figure(figsize=(5, 5))
plt.scatter(df_filter['members_count'], df_filter['views'],color='orange', alpha=0.3)
plt.xlabel('Подписчики')
plt.ylabel('Просмотры')
plt.title('Зависимость просмотров от кол-ва подписчиков')
plt.show()
```

```python
plt.figure(figsize=(15, 5))
plt.subplot(1, 2, 1)
sns.boxplot(x='group_name', y='views', data=df_filter)
plt.xticks(rotation=45, ha='right')
plt.ylabel('Просмотры')
plt.xlabel('Сообщества')
plt.title('Боксплот для просмотров по сообществам')

plt.subplot(1, 2, 2)
sns.boxplot(x='activity', y='views', data=df_filter)
plt.xticks(rotation=45, ha='right')
plt.ylabel('Просмотры')
plt.xlabel('Тип активности')
plt.title('Боксплот для просмотров по типу активности')

plt.show()
```

```python
pd.options.display.float_format = '{:,.0f}'.format

top_5_groups = df_filter.groupby('group_name')['views'].mean().sort_values(ascending=False).head(5)
print("Топ-5 популярных сообществ по средним просмотрам:\n")
print(top_5_groups)

top_3_activities = df_filter.groupby('activity')['views'].mean().sort_values(ascending=False).head(3)
print("\nТоп-3 популярных активности по средним просмотрам:\n")
print(top_3_activities)
```

```python
#смотрим средние просмотры в каждом типе активности в разбивке по длительности роликов
bins = [0, 1800, 3600, 7200, float('inf')]
labels = ['Короткие', 'Средние', 'Длинные', 'Очень длинные']
df_filter['duration_category'] = pd.cut(df_filter['duration'], bins=bins, labels=labels)

for activity in df_filter['activity'].unique():
    df_1 = df_filter[df_filter['activity'] == activity]
    avg_views = df_1.groupby('duration_category')['views'].mean()
    print(f"\nСредние просмотры в активности '{activity}' в разбивке по длительности видео:\n{avg_views}")
```

# Анализ трендов

```python
df_trends = pd.read_csv('../parsing_scraping/videos_trend_pars.csv')
```

```python
from datetime import datetime

dttm_pars = datetime(2025, 3, 4)  # дата парсинга трендов
dttm = dttm_pars - pd.Timedelta(days=5)  # предположительно, в тренды попадают видео за последние 5 дней
```

```python
df_filter['title'] = df_filter['title'].str.lower().str.strip()
df_trends['title'] = df_trends['title'].str.lower().str.strip()

df_last_5 = df_filter[df_filter['date'] >= dttm]

peresech = pd.merge(
    df_last_5[['title', 'views', 'date']],
    df_trends[['title', 'views']], on='title', how='inner', suffixes=('_main', '_trends'))

peresech = peresech.sort_values(by='views_main', ascending=False)

print(f"Найдено пересечений: {len(peresech)}")
peresech.head(10)
```

Нашли два шоу, которые попали в тренды за последние 5 дней, кол-во просмотров 232 тыс и 874 тыс, что относительно хорошие показатели по популярности среди пользователей
